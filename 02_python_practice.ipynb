```python
def get_even_numbers(numbers):
    result = []
    for num in numbers:
        if num % 2 == 0:
            result.append(num)
    return result

print(get_even_numbers([1 , 2 , 3 , 4 , 5 , 6]))
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

print(calculate_average([10 , 20 , 30]))
```

```python
def sum_positive(numbers):
    total = 0
    for num in numbers:
        if num > 0:
            total = total + num
    return total

print(sum_positive([-5 , 10 , -2 , 7 , 3]))
```

```python
def count_positive(numbers):
  count = 0
  for num in numbers:
    if num > 0:
      count = count + 1
  return count

print(count_positive([-2 , 5 , 0 , 8 , -1 , 3]))
```

```python
def find_smallest(numbers):
    smallest = numbers[0]
    for num in numbers:
        if num < smallest:
            smallest = num
    return smallest

print(find_smallest([14 , 6 , 21 , 3 , 9]))
```

```python
def square_numbers(numbers):
    result = []
    for num in numbers:
        result.append(num ** 2)
    return result

print(square_numbers([2 , 3 , 5]))
```

```python
def long_words(words):
    result = []
    for word in words:
        if len(word) > 4:
            result.append(word)
    return result
print(long_words(["cat" , "apple" , "dog" , "mango" , "elephant"]))
```

```python
def unique_numbers(numbers):
    result = []
    for num in numbers:
        if num not in result:
            result.append(num)
    return result

print(unique_numbers([1 , 2 , 2 , 3 , 1 , 4]))
```

```python
def factorial(n):
    result = 1

    for num in range(1 , n + 1):
        result = result * num
    return result

print(factorial(5))
```
